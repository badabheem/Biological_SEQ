```python
from Bio.Seq import Seq
from Bio.SeqUtils import gc_fraction

# Open the FASTA file and process the sequence
with open("/content/GC_contet_Seq.txt", "r") as file:
        overall_gc_count = 0
        overall_length = 0
        for line in file:
                line = line.strip()  # Remove leading/trailing whitespace
                if not line.startswith(">"):  # Skip header lines starting with ">"
                        sequence = Seq(line)  # Create a Seq object for the sequence
                        gc_content = gc_fraction(sequence) * 100  # Calculate GC content percentage

                        print(f"Original Sequence: {sequence}")
                        print(f"GC Content: {gc_content:.2f}%")
                        print("-"*20)

        # After processing all sequences, calculate and print overall GC content
        if overall_length > 0:
               overall_gc_percentage = (overall_gc_count / overall_length) * 100
               print(f"Total GC content for all sequences: {overall_gc_percentage:.2f}%")
        else:
               print("No valid sequence found in the file.")
```

```python
from Bio.Seq import Seq
from Bio.SeqUtils import gc_fraction
from collections import Counter

# DNA sequence
dna = Seq("ATGCGTACGTTAGCGATCGATCGATGCGTACGATCG")
# 1. Sequence length
length = len(dna)
# 2. GC content
gc_content = gc_fraction(dna) * 100
# 3. Nucleotide composition
composition = Counter(dna)
#4. Compliment Seq of DNA
compliment_seq = dna.complement()
print("Original Sequence:", dna)
print("Complement Sequence:", compliment_seq)
#5. Reverse Compliment Seq of DNA
Rev_compliment = dna.reverse_complement()
print("Original Sequence:", dna)
print("Reverse Compliment Sequence:", Rev_compliment)
#6. Transcribe of DNA Seq
rna_seq = dna.transcribe()
print("Original Sequence:", dna)
print("RNA Sequence:", rna_seq)
#7. Translate RNA to Protein Seq
protein_seq = dna.translate()
print("Original Sequence:", dna)
print("RNA Seq:", rna_seq)
print("Protein Sequence:", protein_seq)
# Print the results

print("DNA Sequence:", dna)
print("Sequence Length:", length, "bp")
print("GC Content:", round(gc_content, 2), "%")
print("Nucleotide Composition:")
print("A:", composition["A"])
print("T:", composition["T"])
print("G:", composition["G"])
print("C:", composition["C"])

# Percentage composition
print("\nNucleotide Percentage:")
for nucleotide in "ATGC":
    percentage = (composition[nucleotide] / length) * 100
    print(nucleotide, ":", round(percentage, 2), "%")
```

```python
from Bio.Seq import Seq
from Bio.SeqUtils import gc_fraction

# Open the FASTA file and process the sequence
with open("/content/GC_contet_Seq.txt", "r") as file:
  sequence_data = ''
  for line in file:
    line = line.strip()
    #Skip the header line starting with >
    if not line.startswith(">"):
        sequence_data += line

# Create a Seq object for the accumulated sequence data outside the loop
seq_obj = Seq(sequence_data)
GC_content = gc_fraction(seq_obj) * 100
print("Original Sequence:", seq_obj)
print("GC Content:", round(GC_content, 2), "%")
Count_C = seq_obj.count("C")
Count_G = seq_obj.count("G")
print("Count of C:", Count_C)
print("Count of G:", Count_G)
gc_count = (Count_C + Count_G) / len(seq_obj) * 100
print("Manual GC Count:", round(gc_count, 2), "%")
```
